```python
import pandas as pd
import plotly.express as px

# Cargar el dataset oficial solicitado por los evaluadores
df = pd.read_csv("vehicles_us.csv")

# --- HISTOGRAMA: Distribución de Precios de los Vehículos ---
fig_hist = px.histogram(
    df, x='price', 
    title='Distribución de Precios de los Vehículos',
    labels={'price': 'Precio (USD)'}, template='plotly_dark'
)
fig_hist.show()

# --- SCATTERPLOT: Odómetro (Kilometraje) vs Precio ---
fig_scatter = px.scatter(
    df, x='odometer', y='price', color='condition',
    title='Análisis de Dispersión: Kilometraje (Odometer) vs Precio',
    labels={'odometer': 'Kilometraje', 'price': 'Precio (USD)', 'condition': 'Condición'},
    template='plotly_dark'
)
fig_scatter.show()
```
